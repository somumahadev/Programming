```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def show_image(title, image, cmap=None):
    plt.figure(figsize=(6, 4))
    if image.ndim == 2:
        plt.imshow(image, cmap=cmap or "gray")
    else:
        # Convert BGR to RGB for correct colour display in Matplotlib
        plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.axis("off")
    plt.show()

def read_image(path="image.jpg", grayscale=False):
    flag = cv2.IMREAD_GRAYSCALE if grayscale else cv2.IMREAD_COLOR
    image = cv2.imread(path, flag)
    if image is None:
        raise FileNotFoundError(
            f"Could not read '{path}'. Check the file name and notebook folder."
        )
    return image
```

# Practical 2: RGB to grayscale conversion

**Aim:** To convert a colour image into a grayscale image using weighted intensity values.

### Theory and explanation
Human vision does not respond equally to the three colour channels. The green component contributes more strongly to perceived brightness, followed by red and blue. The weighted conversion therefore gives a more natural grayscale result than a simple average.

**Formula / principle:**  
Gray = 0.299R + 0.587G + 0.114B. OpenCV reads BGR data, so the channel variables must be assigned in the correct order.

### Algorithm
1. Read the colour image.
2. For each pixel, obtain its red, green and blue values.
3. Calculate the weighted sum.
4. Store the result as an 8-bit intensity between 0 and 255.
5. Display the grayscale image.

### Program 1 — Using built-in functions
Run the following cell to perform the operation using library functions.

### Program 2 — Manual implementation
The following version implements the central image-processing operation explicitly.

**Observation:** The manual version uses loops and the weighted formula for every pixel. NumPy is used only to create the output array.

**Result:** The colour image is converted into a single-channel grayscale image.


```python
import cv2

img = cv2.imread("image.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imshow("Grayscale image", gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img = cv2.imread("image.jpg")
h, w = img.shape[:2]
gray = np.zeros((h, w), dtype=np.uint8)

for i in range(h):
    for j in range(w):
        b, g, r = img[i, j]
        value = 0.299*r + 0.587*g + 0.114*b
        gray[i, j] = int(value)

cv2.imshow("Grayscale image", gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Practical 3: Grayscale to binary conversion using thresholding

**Aim:** To convert a grayscale image into a binary image using a fixed threshold.

### Theory and explanation
Thresholding separates pixels into two groups. Pixels above the selected threshold become white and the remaining pixels become black. It is a basic segmentation technique and is useful when foreground and background intensities differ.

**Formula / principle:**  
g(x,y) = 255 if f(x,y) > T; otherwise g(x,y) = 0. Here T is the threshold value.

### Algorithm
1. Read the image in grayscale.
2. Choose a threshold T, such as 127.
3. Compare each pixel with T.
4. Assign 255 when the pixel is greater than T; otherwise assign 0.
5. Display the binary result.

### Program 1 — Using built-in functions
Run the following cell to perform the operation using library functions.

### Program 2 — Manual implementation
The following version implements the central image-processing operation explicitly.

**Observation:** A fixed threshold may not work well when lighting varies across the image. Adaptive thresholding can be studied as a separate exercise.

**Result:** A binary image containing only black and white pixels is obtained.


```python
import cv2

img = cv2.imread("image.jpg", 0)
_, binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

cv2.imshow("Binary image", binary)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img = cv2.imread("image.jpg", 0)
h, w = img.shape
binary = np.zeros((h, w), dtype=np.uint8)

for i in range(h):
    for j in range(w):
        if img[i, j] > 127:
            binary[i, j] = 255
        else:
            binary[i, j] = 0

cv2.imshow("Binary image", binary)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Practical 4: Image negative transformation

**Aim:** To produce the negative of a grayscale image.

### Theory and explanation
The negative transformation reverses the intensity scale. Dark pixels become bright and bright pixels become dark. This can make details in certain image regions easier to inspect.

**Formula / principle:**  
s = L - 1 - r. For an 8-bit image, L = 256, therefore s = 255 - r.

### Algorithm
1. Read the image in grayscale.
2. Visit each pixel.
3. Subtract its intensity from 255.
4. Store and display the transformed pixel.

### Program 1 — Using built-in functions
Run the following cell to perform the operation using library functions.

### Program 2 — Manual implementation
The following version implements the central image-processing operation explicitly.

**Observation:** The transformation is applied independently to every pixel and does not change the image dimensions.

**Result:** The negative version of the grayscale image is displayed.


```python
import cv2

img = cv2.imread("image.jpg", 0)
negative = cv2.bitwise_not(img)

cv2.imshow("Negative image", negative)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img = cv2.imread("image.jpg", 0)
h, w = img.shape
negative = np.zeros((h, w), dtype=np.uint8)

for i in range(h):
    for j in range(w):
        negative[i, j] = 255 - int(img[i, j])

cv2.imshow("Negative image", negative)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Practical 5: Brightness adjustment

**Aim:** To increase or decrease image brightness by adding a constant value to pixel intensities.

### Theory and explanation
Brightness adjustment shifts pixel intensities. A positive constant makes the image brighter; a negative constant makes it darker. Values must be clipped to the valid 8-bit range.

**Formula / principle:**  
g(x,y) = clip(f(x,y) + C, 0, 255).

### Algorithm
1. Read the colour image.
2. Choose a brightness offset C.
3. Add C to each channel value.
4. Clip values below 0 to 0 and above 255 to 255.
5. Display the adjusted image.

### Program 1 — Using built-in functions
Run the following cell to perform the operation using library functions.

### Program 2 — Manual implementation
The following version implements the central image-processing operation explicitly.

**Observation:** Change +50 to -50 to darken the image. Clipping prevents invalid intensity values.

**Result:** The image brightness is increased by the selected offset.


```python
import cv2

img = cv2.imread("image.jpg")
bright = cv2.convertScaleAbs(img, alpha=1, beta=50)

cv2.imshow("Brighter image", bright)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img = cv2.imread("image.jpg")
h, w, ch = img.shape
result = np.zeros_like(img)

for i in range(h):
    for j in range(w):
        for k in range(ch):
            value = int(img[i, j, k]) + 50
            result[i, j, k] = min(255, max(0, value))

cv2.imshow("Brighter image", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Practical 6: Contrast adjustment

**Aim:** To modify image contrast by multiplying pixel intensities by a scale factor.

### Theory and explanation
Contrast describes the difference between light and dark regions. A scale factor greater than one spreads intensities toward the extremes, while a factor between zero and one reduces the contrast.

**Formula / principle:**  
g(x,y) = clip(αf(x,y), 0, 255), where α is the contrast scale factor.

### Algorithm
1. Read the image.
2. Select a contrast factor α.
3. Multiply each channel value by α.
4. Clip the result to the 0–255 range.
5. Display the result.

### Program 1 — Using built-in functions
Run the following cell to perform the operation using library functions.

### Program 2 — Manual implementation
The following version implements the central image-processing operation explicitly.

**Observation:** This is a simple contrast scaling operation. More advanced contrast enhancement includes histogram equalization and contrast stretching.

**Result:** The contrast of the input image is modified.


```python
import cv2

img = cv2.imread("image.jpg")
contrast = cv2.convertScaleAbs(img, alpha=1.5, beta=0)

cv2.imshow("Contrast adjusted", contrast)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img = cv2.imread("image.jpg")
h, w, ch = img.shape
result = np.zeros_like(img)

for i in range(h):
    for j in range(w):
        for k in range(ch):
            value = int(1.5 * int(img[i, j, k]))
            result[i, j, k] = min(255, value)

cv2.imshow("Contrast adjusted", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Practical 7: Image resizing using nearest-neighbour mapping

**Aim:** To change the dimensions of an image.

### Theory and explanation
Resizing maps pixels from the source image to a new grid. Nearest-neighbour interpolation selects the closest source pixel and is simple to implement, although it may produce blocky results when enlarging an image.

**Formula / principle:**  
src_y = floor(dst_y × H / new_H), src_x = floor(dst_x × W / new_W).

### Algorithm
1. Read the input image and record its dimensions.
2. Specify the required output width and height.
3. For each output coordinate, calculate the corresponding source coordinate.
4. Copy the source pixel to the output image.
5. Display the resized image.

### Program 1 — Using built-in functions
Run the following cell to perform the operation using library functions.

### Program 2 — Manual implementation
The following version implements the central image-processing operation explicitly.

**Observation:** The manual implementation uses nearest-neighbour interpolation so that it matches the selected OpenCV interpolation method.

**Result:** The input image is resized to 300 × 300 pixels.


```python
import cv2

img = cv2.imread("image.jpg")
resized = cv2.resize(img, (300, 300),
                     interpolation=cv2.INTER_NEAREST)

cv2.imshow("Resized image", resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img = cv2.imread("image.jpg")
h, w = img.shape[:2]
new_h, new_w = 300, 300
resized = np.zeros((new_h, new_w, 3), dtype=np.uint8)

for i in range(new_h):
    for j in range(new_w):
        src_y = int(i * h / new_h)
        src_x = int(j * w / new_w)
        resized[i, j] = img[src_y, src_x]

cv2.imshow("Resized image", resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Practical 8: Image rotation by 90 degrees

**Aim:** To rotate an image by 90 degrees in the clockwise direction.

### Theory and explanation
Rotation changes the spatial position of pixels. A 90-degree clockwise rotation swaps the image dimensions: the output height becomes the original width and the output width becomes the original height.

**Formula / principle:**  
For an image of height H, source pixel (i,j) moves to output coordinate (j, H−1−i).

### Algorithm
1. Read the input image.
2. Obtain its height and width.
3. Create an output array of size width × height.
4. Map each source pixel to its clockwise-rotated position.
5. Display the rotated image.

### Program 1 — Using built-in functions
Run the following cell to perform the operation using library functions.

### Program 2 — Manual implementation
The following version implements the central image-processing operation explicitly.

**Observation:** The manual coordinate mapping performs an exact 90-degree rotation without interpolation.

**Result:** The image is rotated 90 degrees clockwise.


```python
import cv2

img = cv2.imread("image.jpg")
rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

cv2.imshow("Rotated image", rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img = cv2.imread("image.jpg")
h, w = img.shape[:2]
rotated = np.zeros((w, h, 3), dtype=np.uint8)

for i in range(h):
    for j in range(w):
        rotated[j, h - 1 - i] = img[i, j]

cv2.imshow("Rotated image", rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Practical 9: Image translation

**Aim:** To shift an image horizontally and vertically by a specified number of pixels.

### Theory and explanation
Translation moves image content without changing its shape or orientation. Pixels that move outside the output boundary are discarded, and newly exposed areas are filled with black.

**Formula / principle:**  
x' = x + t_x; y' = y + t_y.

### Algorithm
1. Read the input image and obtain its dimensions.
2. Choose horizontal and vertical shifts.
3. Create a blank output image.
4. Map each source pixel to its translated coordinate if it lies inside the output boundary.
5. Display the translated image.

### Program 1 — Using built-in functions
Run the following cell to perform the operation using library functions.

### Program 2 — Manual implementation
The following version implements the central image-processing operation explicitly.

**Observation:** The transformation matrix and manual mapping use the same positive horizontal and vertical shifts.

**Result:** The image content is shifted 50 pixels right and 30 pixels down.


```python
import cv2
import numpy as np

img = cv2.imread("image.jpg")
h, w = img.shape[:2]
M = np.float32([[1, 0, 50], [0, 1, 30]])
translated = cv2.warpAffine(img, M, (w, h))

cv2.imshow("Translated image", translated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img = cv2.imread("image.jpg")
h, w = img.shape[:2]
translated = np.zeros_like(img)

tx, ty = 50, 30

for y in range(h):
    for x in range(w):
        new_x = x + tx
        new_y = y + ty
        if 0 <= new_x < w and 0 <= new_y < h:
            translated[new_y, new_x] = img[y, x]

cv2.imshow("Translated image", translated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Practical 11: Grayscale image histogram

**Aim:** To calculate and plot the intensity histogram of a grayscale image.

### Theory and explanation
A histogram counts how many pixels have each intensity value. For an 8-bit grayscale image there are 256 possible intensity levels, from 0 (black) to 255 (white). The histogram gives a summary of the image's tonal distribution.

**Formula / principle:**  
h(r_k) = n_k, where n_k is the number of pixels having intensity r_k.

### Algorithm
1. Read the image in grayscale.
2. Create 256 counters initialized to zero.
3. For every pixel, increment the counter corresponding to its intensity.
4. Plot intensity level against frequency.

### Program 1 — Using built-in functions
Run the following cell to perform the operation using library functions.

### Program 2 — Manual implementation
The following version implements the central image-processing operation explicitly.

**Observation:** The manual version performs the frequency counting explicitly. Matplotlib is used only to draw the graph.

**Result:** The histogram showing the frequency of each grayscale intensity is plotted.


```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("image.jpg", 0)
hist = cv2.calcHist([img], [0], None, [256], [0, 256])

plt.plot(hist)
plt.xlabel("Intensity")
plt.ylabel("Number of pixels")
plt.title("Grayscale histogram")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("image.jpg", 0)
h, w = img.shape
hist = [0] * 256

for y in range(h):
    for x in range(w):
        intensity = int(img[y, x])
        hist[intensity] += 1

plt.plot(range(256), hist)
plt.xlabel("Intensity")
plt.ylabel("Number of pixels")
plt.title("Grayscale histogram")
plt.show()
```

# Practical 12: Histogram equalization

**Aim:** To improve the global contrast of a grayscale image using histogram equalization.

### Theory and explanation
Histogram equalization redistributes intensity values using the cumulative distribution function (CDF). It can reveal details in images whose intensities occupy only a narrow range. The result depends on the original image histogram.

**Formula / principle:**  
s_k = round(255 × (CDF(k) − CDF_min) / (MN − CDF_min)), where M×N is the number of pixels.

### Algorithm
1. Read the image in grayscale.
2. Calculate the 256-bin histogram.
3. Compute the cumulative histogram.
4. Normalize the CDF to the range 0–255.
5. Replace each pixel using the mapping and display the result.

### Program 1 — Using built-in functions
Run the following cell to perform the operation using library functions.

### Program 2 — Manual implementation
The following version implements the central image-processing operation explicitly.

**Observation:** If the image is constant (all pixels have the same value), the denominator can be zero. In that special case, retain the original image rather than applying the mapping.

**Result:** The grayscale image is contrast-enhanced through histogram equalization.


```python
import cv2

img = cv2.imread("image.jpg", 0)
equalized = cv2.equalizeHist(img)

cv2.imshow("Original", img)
cv2.imshow("Equalized", equalized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img = cv2.imread("image.jpg", 0)
h, w = img.shape
hist = [0] * 256

for y in range(h):
    for x in range(w):
        hist[int(img[y, x])] += 1

cdf = []
total = 0
for count in hist:
    total += count
    cdf.append(total)

cdf_min = next(v for v in cdf if v > 0)
mapping = [0] * 256
pixels = h * w

for k in range(256):
    mapping[k] = round(
        (cdf[k] - cdf_min) * 255 / (pixels - cdf_min)
    )

result = np.zeros_like(img)
for y in range(h):
    for x in range(w):
        result[y, x] = mapping[int(img[y, x])]

cv2.imshow("Equalized image", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Practical 13: Average filtering

**Aim:** To smooth an image using a 3 × 3 averaging filter.

### Theory and explanation
An averaging filter replaces each pixel with the mean of the pixels in its neighbourhood. It reduces small intensity variations and can suppress mild noise, but it also blurs edges and fine details.

**Formula / principle:**  
g(x,y) = (1/9) × sum of the nine pixels in the 3 × 3 neighbourhood.

### Algorithm
1. Read the image in grayscale.
2. Visit each pixel that has a complete 3 × 3 neighbourhood.
3. Add the nine neighbourhood values.
4. Divide the sum by nine and assign the mean to the output pixel.
5. Display the smoothed image.

### Program 1 — Using built-in functions
Run the following cell to perform the operation using library functions.

### Program 2 — Manual implementation
The following version implements the central image-processing operation explicitly.

**Observation:** The manual code leaves the one-pixel border black because a complete 3 × 3 neighbourhood is unavailable there. Border-handling methods can be added for a more exact comparison.

**Result:** The image is smoothed using a 3 × 3 mean filter.


```python
import cv2

img = cv2.imread("image.jpg", 0)
blur = cv2.blur(img, (3, 3))

cv2.imshow("Average filtered image", blur)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img = cv2.imread("image.jpg", 0)
h, w = img.shape
result = np.zeros_like(img)

for y in range(1, h-1):
    for x in range(1, w-1):
        total = 0
        for dy in range(-1, 2):
            for dx in range(-1, 2):
                total += int(img[y+dy, x+dx])
        result[y, x] = total // 9

cv2.imshow("Average filtered image", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Practical 14: Median filtering

**Aim:** To remove impulse noise from an image using a 3 × 3 median filter.

### Theory and explanation
A median filter sorts the values in a local neighbourhood and selects the middle value. Unlike averaging, it is less influenced by extreme values and is particularly useful for reducing salt-and-pepper noise.

**Formula / principle:**  
g(x,y) = median{f(x+i,y+j) : i,j ∈ {−1,0,1}}.

### Algorithm
1. Read the image in grayscale.
2. Collect the nine values in each 3 × 3 neighbourhood.
3. Sort the values.
4. Select the fifth value (the median).
5. Store the result and display the filtered image.

### Program 1 — Using built-in functions
Run the following cell to perform the operation using library functions.

### Program 2 — Manual implementation
The following version implements the central image-processing operation explicitly.

**Observation:** A 3 × 3 window contains nine values, so the median is the value at sorted index 4. The border is not processed in this basic implementation.

**Result:** Impulse noise is reduced using median filtering.


```python
import cv2

img = cv2.imread("image.jpg", 0)
median = cv2.medianBlur(img, 3)

cv2.imshow("Median filtered image", median)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Practical 15: Sobel edge detection

**Aim:** To detect horizontal and vertical intensity changes using the Sobel operator.

### Theory and explanation
The Sobel operator estimates the image gradient using two 3 × 3 convolution masks. Large gradient magnitudes indicate likely edges. The x and y responses can be combined to estimate overall edge strength.

**Formula / principle:**  
Gx = I * Kx; Gy = I * Ky; G = sqrt(Gx² + Gy²).

### Algorithm
1. Read the image in grayscale.
2. Define the Sobel x and y kernels.
3. For each interior pixel, multiply the neighbourhood by each kernel and sum the results.
4. Calculate the gradient magnitude.
5. Clip the magnitude to 255 and display it.

### Program 1 — Using built-in functions
Run the following cell to perform the operation using library functions.

### Program 2 — Manual implementation
The following version implements the central image-processing operation explicitly.

**Observation:** The two kernels estimate changes along the x and y directions. The combined magnitude is a simple edge-strength measure.

**Result:** Edges and strong intensity transitions are highlighted.


```python
import cv2
import numpy as np

img = cv2.imread("image.jpg", 0)
gx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
gy = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

magnitude = cv2.magnitude(gx, gy)
result = cv2.convertScaleAbs(magnitude)

cv2.imshow("Sobel edges", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img = cv2.imread("image.jpg", 0)
h, w = img.shape
result = np.zeros_like(img)

kx = [[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]]
ky = [[-1, -2, -1], [0, 0, 0], [1, 2, 1]]

for y in range(1, h-1):
    for x in range(1, w-1):
        sx, sy = 0, 0
        for a in range(3):
            for b in range(3):
                p = int(img[y+a-1, x+b-1])
                sx += p * kx[a][b]
                sy += p * ky[a][b]
        magnitude = int((sx*sx + sy*sy) ** 0.5)
        result[y, x] = min(255, magnitude)

cv2.imshow("Sobel edges", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Practical 17: Gaussian filtering

**Aim:** To smooth an image using a Gaussian-weighted neighbourhood.

### Theory and explanation
Gaussian filtering assigns greater weight to pixels near the centre of the kernel and smaller weights to distant pixels. It reduces high-frequency noise while generally preserving edges better than a simple box average of similar size.

**Formula / principle:**  
G(x,y) = (1/(2πσ²)) exp(−(x²+y²)/(2σ²)); the kernel is normalized so its weights sum to one.

### Algorithm
1. Read the image in grayscale.
2. Create a normalized 3 × 3 Gaussian kernel.
3. Multiply each neighbourhood pixel by its corresponding kernel weight.
4. Sum the weighted values.
5. Store and display the filtered image.

### Program 1 — Using built-in functions
Run the following cell to perform the operation using library functions.

### Program 2 — Manual implementation
The following version implements the central image-processing operation explicitly.

**Observation:** The manual version uses the common 3 × 3 approximation with weights summing to 16. It is a discrete Gaussian-like kernel.

**Result:** The image is smoothed using Gaussian-weighted averaging.


```python
import cv2

img = cv2.imread("image.jpg", 0)
blur = cv2.GaussianBlur(img, (3, 3), 0)

cv2.imshow("Gaussian filtered image", blur)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img = cv2.imread("image.jpg", 0)
h, w = img.shape
result = np.zeros_like(img)

kernel = [[1, 2, 1],
          [2, 4, 2],
          [1, 2, 1]]
weight_sum = 16

for y in range(1, h-1):
    for x in range(1, w-1):
        total = 0
        for a in range(3):
            for b in range(3):
                total += int(img[y+a-1, x+b-1]) * kernel[a][b]
        result[y, x] = total // weight_sum

cv2.imshow("Gaussian filtered image", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Practical 24: Morphological opening and closing

**Aim:** To perform opening and closing operations on a binary image.

### Theory and explanation
Opening is erosion followed by dilation. It can remove small foreground objects while retaining the general shape of larger objects. Closing is dilation followed by erosion and can fill small holes or gaps in foreground regions.

**Formula / principle:**  
Opening: A ○ B = (A ⊖ B) ⊕ B. Closing: A • B = (A ⊕ B) ⊖ B.

### Algorithm
1. Read the binary image and define a 3 × 3 structuring element.
2. For opening, perform erosion followed by dilation.
3. For closing, perform dilation followed by erosion.
4. Display both results.

### Program 1 — Using built-in functions
Run the following cell to perform the operation using library functions.

### Program 2 — Manual implementation
The following version implements the central image-processing operation explicitly.

**Observation:** The helper functions implement the same full 3 × 3 neighbourhood rules used in the preceding erosion and dilation practicals. Border pixels are left black in this simple version.

**Result:** Opening and closing results are produced from the binary input image.


```python
import cv2
import numpy as np

img = cv2.imread("binary.jpg", 0)
kernel = np.ones((3, 3), np.uint8)

opening = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)
closing = cv2.morphologyEx(img, cv2.MORPH_CLOSE, kernel)

cv2.imshow("Opening", opening)
cv2.imshow("Closing", closing)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

img = cv2.imread("binary.jpg", 0)
h, w = img.shape

def erode(src):
    dst = np.zeros_like(src)
    for y in range(1, h-1):
        for x in range(1, w-1):
            if all(src[y+dy, x+dx] > 0
                   for dy in range(-1, 2)
                   for dx in range(-1, 2)):
                dst[y, x] = 255
    return dst

def dilate(src):
    dst = np.zeros_like(src)
    for y in range(1, h-1):
        for x in range(1, w-1):
            if any(src[y+dy, x+dx] > 0
                   for dy in range(-1, 2)
                   for dx in range(-1, 2)):
                dst[y, x] = 255
    return dst

opening = dilate(erode(img))
closing = erode(dilate(img))

cv2.imshow("Opening", opening)
cv2.imshow("Closing", closing)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
